# Predict 

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
X=[[1],[2],[3],[4]]
y = [2, 4, 6, 8]
model.fit(X, y)
user_value=float(input("Enter a number to predict:"))
prediction=model.predict([[user_value]])
print("Prediction for input 10 is!", prediction)
```

# Same Code but this time by taking input from the user

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
X=[[1],[2],[3],[4]]
y = [2, 4, 6, 8]
model.fit(X, y)
user_value=float(input("Enter a number to predict:"))
prediction=model.predict([[user_value]])
print("Prediction for input 10 is!", prediction)
```

# Visualization The Prediction

```python
import matplotlib.pyplot as plt
plt.scatter(X,y,color='orange' ,label =' Data Point')

# Reggression line

plt.plot(X,model.predict(X), color="red" , label ='Regression Line')

plt.legend(loc='center left', bbox_to_anchor=(1,0.5))
plt.title("linear Regression line")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout() 
plt.show()
```

# Activity 1: Predict the price of the banna

```python
weights=np.array([1,2,3,4,5]).reshape(-1,1)
prices=np.array([30,60,90,120,150])
model.fit(weights,prices)
prediction= model.predict([[10]])
print("Prediction for input 10 is!", prediction)
plt.scatter(weights,prices,color='orange' ,label =' Data Point')
plt.plot(weights,model.predict(weights), color="red" , label ='Regression Line')

plt.legend(loc='center left', bbox_to_anchor=(1,0.5))
plt.title("linear Regression line")
plt.xlabel("weights")
plt.ylabel("prices")
plt.tight_layout() 
plt.show()
```
